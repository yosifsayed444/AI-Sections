## Common container functions

Demonstrations of `len`, `sum`, `min`, `max`, `sorted`, `reversed`, `any`, `all`, `map`, `filter`, `zip`, and `dict`/`set` comprehensions.

```python
lst = [3, 1, 4, 1, 5]
tpl = (10, 20, 5)
s = {2, 3, 7}
d = {'a': 3, 'b': 1}

print('len(lst) =', len(lst))
print('sum(lst) =', sum(lst))
print('min(tpl) =', min(tpl))
print('max(tpl) =', max(tpl))
print('sorted(lst) =', sorted(lst))
print('list(reversed(lst)) =', list(reversed(lst)))
print('any(x>3 for x in lst) =', any(x>3 for x in lst))
print('all(x>0 for x in lst) =', all(x>0 for x in lst))
print('map square =', list(map(lambda x: x*x, lst)))
print('filter odd =', list(filter(lambda x: x%2==1, lst)))
print('zip(lst, tpl) =', list(zip(lst, tpl)))
print('dict comprehension example:', {k: v*2 for k, v in d.items()})
print('set comprehension example:', {x*x for x in lst})
```

## Break / continue / else — detailed illustrations

The `else` block on loops runs only when a loop terminates normally (no `break`). `continue` does not prevent `else` from running. Here are multiple examples showing behavior and edge cases.

```python
# Example 1: for loop completes normally -> else runs
for i in range(3):
    print('ex1 i=', i)
else:
    print('ex1 else: loop completed without break')

# Example 2: break prevents else
for i in range(5):
    print('ex2 i=', i)
    if i == 2:
        print('ex2 breaking at', i)
        break
else:
    print('ex2 else: will not run')

# Example 3: continue skips iteration but does not stop else
for i in range(4):
    if i % 2 == 0:
        print('ex3 even, continue', i)
        continue
    print('ex3 odd', i)
else:
    print('ex3 else: loop completed without break')

# Edge case: return inside loop (in function) prevents else from running because function exits
def fn_with_return():
    for i in range(3):
        if i == 1:
            return 'returned early'
    else:
        return 'completed normally'

print('fn_with_return ->', fn_with_return())

# Edge case: exception in loop prevents else
try:
    for i in range(3):
        if i == 1:
            raise ValueError('boom')
    else:
        print('this else will not run because exception occurs')
except Exception as e:
    print('caught exception, else did not run:', e)
```

## Parameter types (positional-only, positional-or-keyword, keyword-only) and pitfalls

We'll show signatures and common pitfalls including mutable default arguments.

```python
main_lst = []
def append_good(value, lst=None):
    if lst is None:
        lst = []
    lst.append(value)
    return lst

print('append_good first:', append_good(1 , main_lst))
print('append_good second:', append_good(2 , main_lst))
```

```python
# positional-only parameters (Python 3.8+)
def pos_only(a, b, /, c):
    return a + b + c
print('pos_only(1,2,3)=', pos_only(1,2,3))
# trying keyword for positional-only would raise TypeError if uncommented
# pos_only(a=1, b=2, c=3)

# positional-or-keyword (normal)
def pos_or_kw(a, b=10):
    return a * b
print('pos_or_kw(2)=', pos_or_kw(2))
print('pos_or_kw(2,5)=', pos_or_kw(2,5))
print('pos_or_kw(a=3, b=4)=', pos_or_kw(a=3, b=4))

# keyword-only parameters: use * to mark end of positional args
def kw_only(a, *, verbose=False):
    if verbose:
        print('verbose on')
    return a
print('kw_only(5, verbose=True)=', kw_only(5, verbose=True))

# Pitfall: mutable default argument
def append_bad(value, lst=[]):
    lst.append(value)
    return lst

print('append_bad first:', append_bad(1))
print('append_bad second:', append_bad(2))  # retains state across calls

# Safe pattern:
def append_good(value, lst=None):
    if lst is None:
        lst = []
    lst.append(value)
    return lst

print('append_good first:', append_good(1))
print('append_good second:', append_good(2))
```

## Variable-parameter functions: `*args` and `**kwargs`

Examples of collecting extra args/kwargs, and forwarding them to other functions.

```python
def combined(a, *args, sep='-', **kwargs):
    print('a =', a)
    print('*args =', args)
    print('sep =', sep)
    print('**kwargs =', kwargs)

combined(1, 2, 3, sep='|', x=10, y=20)

# forwarding / unpacking example
def wrapper(func, *args, **kwargs):
    return func(*args, **kwargs)

def greet(name, punctuation='!'):
    return f'Hello {name}{punctuation}'

print(wrapper(greet, 'Alice'))
print(wrapper(greet, 'Bob', punctuation='.') )

# unpacking when calling
nums = (1, 2)
params = {'sep': ','}
print('unpack *nums ->', *nums)
print('unpack **params used in function:')
def show_sep(a, sep=','):
    print(a, sep=sep)
show_sep('a b', **{'sep': ';'})
```

## Lambda expressions

Short anonymous functions useful for small callbacks. Examples with `sorted`, `map`, and `filter`.

```python
square = lambda x: x*x
print('square(5)=', square(5))
nums = [5, 2, 9, 1]
print(type(nums))
print('sorted asc =', sorted(nums))
print('sorted by negative =', sorted(nums, key=lambda x: -x))
print('map cubes =', list(map(lambda x: x**3, nums)))
print('filter even =', list(filter(lambda x: x%2==0, nums)))
```

## Exercises (runnable)

Small exercises with solutions included.

```python
# Exercise 1: average of arbitrary args
def average(*args):
    if not args:
        return None
    return sum(args) / len(args)
print('avg =', average(1,2,3,4))

# Exercise 2: zip to dict
keys = ['a','b','c']
values = [1,2,3]
print('zipped dict =', dict(zip(keys, values)))

# Exercise 3: fibonacci generator
def fib(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a+b
print('first 10 fib =', list(fib(10)))
```
