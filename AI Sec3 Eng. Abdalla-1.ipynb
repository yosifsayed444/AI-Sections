## Variable Arguments (*args) and Argument Unpacking

This section demonstrates two important Python concepts:

1. **`*args` - Variable-length positional arguments**: Allows functions to accept any number of positional arguments
2. **Argument unpacking with `*`**: Expands sequences (lists, tuples) into separate function arguments

The examples below show:
- How `*args` collects multiple arguments into a tuple
- How `*` unpacks a sequence when calling functions
- Practical use cases for both patterns

````python
# The *args parameter collects any number of positional 
# arguments into a tuple called args
def print_args(*args):  
    for arg in args:  
        print (arg)

# arguements get packed into args tuple
print_args("X",1,2)
````

````python
data = ['Jan', 'Feb', 'March', 'April']
# A regular function expecting exactly 4 positional arguments
def test(arg1,arg2,arg3,arg4):  
    print (arg1)
    print (arg2)
    print (arg3)
    print (arg4)

# Unpack the list into separate arguments using *
test(*data)
# print(*data)  # This will print each month separately
````

````python
def mul(*args):
    x=1
    for item in args:
        x*=item
    return x

print(mul(1,2,3))
print(mul(1,2,3,4))
print(mul())
````

## Keyword Arguments (**kwargs) and Dictionary Unpacking

This section demonstrates keyword argument handling in Python:

1. **`**kwargs` - Variable-length keyword arguments**: Allows functions to accept any number of keyword arguments, collected into a dictionary
2. **Dictionary unpacking with `**`**: Expands dictionaries into keyword arguments when calling functions

The examples below show:
- How `**kwargs` collects keyword arguments into a dictionary
- How `**` unpacks a dictionary when calling functions
- The importance of matching parameter names when unpacking dictionaries
- Common errors and how to avoid them

````python
def print_kwargs(**kwargs):  
    for key, value in kwargs.items(): 
        print ("%s : %s" %(key, value)) 
  

print_kwargs(first ='X', mid ='Y', last='Z') 

def print_kwargs(arg1,arg2,arg3): 
    print(arg1)
    print(arg2)
    print(arg3)
  
d = {"arg1" :'X', "arg2" :'Y', "arg3":'Z'}

# # Dictionary keys must match function parameter names when unpacking
# d = {"x" :'xx', "arg2" :'yy', "arg3":'zz'} ERROR
print_kwargs(**d)
````

### Q3: Flexible logging function
Create a function `log(level, *messages, **options)` that:
1. Takes a required `level` parameter
2. Takes any number of message strings via `*messages`
3. Takes optional settings via `**options` (like `timestamp=True`, `file="log.txt"`)
4. Prints: `[LEVEL] message1 message2` and shows any options

```python
def log(level, *messages, **options):
    # Your code here
    pass

# Test cases
log("INFO", "System started")
log("ERROR", "Database", "connection", "failed", timestamp=True)
log("DEBUG", "Processing user", user_id=123, verbose=True)

````python
def log(level, *messages, **options):
    # Combine messages
    combined_message = " ".join(messages)
    
    # Create base log entry
    log_entry = f"[{level}] {combined_message}"
    
    # Add options if provided
    if options:
        options_str = ", ".join([f"{k}={v}" for k, v in options.items()])
        log_entry += f" | Options: {options_str}"
    
    print(log_entry)

# Test cases
log("INFO", "System started")
# [INFO] System started

log("ERROR", "Database", "connection", "failed", timestamp=True)
# [ERROR] Database connection failed | Options: timestamp=True

log("DEBUG", "Processing user", user_id=123, verbose=True)
# [DEBUG] Processing user | Options: user_id=123, verbose=True
````

### Q5: Database query builder
Create a function `query_users(**filters)` that builds a SQL-like query string from keyword arguments.

```python
def query_users(**filters):
    # Your code here - build a query like:
    # "SELECT * FROM users WHERE condition1 AND condition2"
    pass

# Test cases
print(query_users())  # Should return basic SELECT
print(query_users(age=25))  # WHERE age = 25
print(query_users(name="Alice", city="NYC", active=True))

````python
def query_users(**filters):
    base_query = "SELECT * FROM users"
    
    if not filters:
        return base_query
    
    # Build WHERE conditions
    conditions = []
    for key, value in filters.items():
        # Handle string values with quotes
        if isinstance(value, str):
            conditions.append(f"{key} = '{value}'")
        else:
            conditions.append(f"{key} = {value}")
    
    where_clause = " AND ".join(conditions)
    return f"{base_query} WHERE {where_clause}"

# Test cases
print(query_users())  
# SELECT * FROM users

print(query_users(age=25))  
# SELECT * FROM users WHERE age = 25

print(query_users(name="Alice", city="NYC", active=True))
# SELECT * FROM users WHERE name = 'Alice' AND city = 'NYC' AND active = True
````

# Stack Implementation #1

````python
MAX = 10
````

````python
class Stack:
    
    def __init__(self): # constructor
        self.stack = list()
    
    def push(self, item):
        if item not in self.stack:
            self.stack.append(item)
            return 1
        return 0
    
    def pop(self):
        if len(self.stack) > 0:
            return self.stack.pop()
        return ("Stack is empty")
    
    def size(self):
        return len(self.stack)
    
    def isEmpty(self):
        return 1 if len(self.stack) == 0 else 0
    
    def isFull(self):
        return 1 if len(self.stack) == MAX else 0
    
    def clear(self):
        self.stack = list()
````

````python
stack = Stack()
print("Push into stacks:")
print(stack.push(1))
print(stack.push(2))
print(stack.push(3))
print(stack.push(2))
print("Get the size of stack:")
print(stack.size())
print("Is the stack full?")
print(stack.isFull())
print("Pop all items from stack:")
print(stack.pop())
print(stack.pop())
print(stack.pop())
print("Is the stack empty?")
print(stack.isEmpty())
````

# Stack Implementation #2

````python
class Stack:
    
    def __init__(self):
        self.stack = list()
        self.maxSize = 10
        self.top = 0
    
    def push(self,data):
        if self.top >= self.maxSize:
            return 0
        self.stack.append(data)
        self.top += 1
        return 1
        
    def pop(self):
        if self.top <= 0:
            return ("Stack Empty!")
        self.top -= 1
        return self.stack.pop()
        
    def size(self):
        return self.top
````

````python
stack = Stack()
print("Push into stack:")
print(stack.push(1))
print(stack.push(2))
print(stack.push(3))
print("Size of stack:")
print(stack.size())
print("Pop all items from stack:")
print(stack.pop())
print(stack.pop())
print(stack.pop())
print("Size of stack:")
print(stack.size())
````

## Binary Tree Implementation

This section implements a basic **Binary Tree** data structure in Python. A binary tree is a hierarchical data structure where each node has at most two children, referred to as the left child and right child.

### Key Features:
- **Node Structure**: Each node stores a value and references to left and right children
- **Flexible Insertion**: Methods to insert nodes on left or right sides
- **Tree Traversal**: Pre-order traversal implementation (Root → Left → Right)

### Important Notes:
- This is a **general binary tree**, not a Binary Search Tree (no value ordering)
- The insertion methods create **chains** rather than balanced trees
- Demonstrates fundamental tree concepts and recursive traversal

The implementation shows:
- Basic tree node creation and linking
- Recursive tree traversal algorithms
- How tree structures can become unbalanced with certain insertion patterns

````python
class BinaryTree:
    
    def __init__(self, value):
        self.value = value
        self.left_child = None
        self.right_child = None
    
    def insert_left(self, value):
        if self.left_child == None:
            self.left_child = BinaryTree(value)
        else:
            new_node = BinaryTree(value)
            new_node.left_child = self.left_child
            self.left_child = new_node
    
    def insert_right(self, value):
        if self.right_child == None:
            self.right_child = BinaryTree(value)
        else:
            new_node = BinaryTree(value)
            new_node.right_child = self.right_child
            self.right_child = new_node
    
    def pre_order(self): # visit the node then move to the left and backtrack to the right
        print(self.value)

        if self.left_child:
            self.left_child.pre_order()

        if self.right_child:
            self.right_child.pre_order()
    
    def in_order(self):
        pass
    
    def post_order(self):
        pass
````

# Binary Search Tree Implementation

````python
class BinarySearchTree:
    def __init__(self, value):
        self.value = value
        self.left_child = None
        self.right_child = None

    def insert_node(self, value):
        '''
            special case here is when have value equal to root will be inserted in left_tree
        '''
        # case 1 is `value` less than root and tree has left (need to insert in left_tree)
        if value <= self.value and self.left_child:
            self.left_child.insert_node(value)
        # case 2 no left_tree and `value` will be the first left_child
        elif value <= self.value:
            self.left_child = BinarySearchTree(value)
        # case 3 is `value` greater than root and tree has right (need to insert in right_tree)
        elif value > self.value and self.right_child:
            self.right_child.insert_node(value)
        # case 4 no right_tree and `value` will be the first right_child
        else:
            self.right_child = BinarySearchTree(value)
    
    def find_node(self, value):
        if value < self.value and self.left_child:
            return self.left_child.find_node(value)
        if value > self.value and self.right_child:
            return self.right_child.find_node(value)

        return value == self.value
    
    def pre_order(self):
        print(self.value)

        if self.left_child:
            self.left_child.pre_order()

        if self.right_child:
            self.right_child.pre_order()
            
    def clear_node(self):
        self.value = None
        self.left_child = None
        self.right_child = None
        
    def find_minimum_value(self):
        if self.left_child:
            return self.left_child.find_minimum_value()
        else:
            return self.value

    def remove(self, value):
        pass
        
````
